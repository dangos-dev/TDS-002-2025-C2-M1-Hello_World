1. Definir las constantes y colores

```python
default_message: str = "Hello world"
```

3. Lógica de entrada del usuario y el mensaje

```python
try:
    user_name = input("¿Cuál es tu nombre? -> ")

    if user_name:
        print(f"{default_message} {user_name}")
    else:
        print("\n(No se detectó entrada de datos)")
        print(f"{default_message} Nuestro nombre")

except KeyboardInterrupt:
    print("\nEntrada de datos interrumpida por el usuario")
except EOFError:
    print("\nNo se detectó entrada de datos")
```
